```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../data/bank-full.csv", sep = ";")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
for col in df.select_dtypes(include="object").columns:
    print("/n", col)
    print(df[col].value_counts())
```

```python
for col in ['job', 'education', 'contact', 'poutcome']:
    print(f"/n{col}")
    print(df[col].value_counts(normalize=True)*100)
```

```python
df['age_group'] = pd.cut(
    df['age'],
    bins = [18, 30, 40, 50, 60, 100],
    labels = [
        '18-30',
        '31-40',
        '41-50',
        '51-60',
        '60+'
    ]
)
```

```python
df['balance_segment'] = pd.cut(
    df['balance'],
    bins=[-10000, 0, 1000, 5000, 100000],
    labels= [
        'Negative',
        'Low',
        'Medium',
        'High'
    ]
)
```

```python
df['campaign_group'] = pd.cut(
    df['campaign'],
    bins=[0, 2, 5, 10, 100],
    labels=[
        'Low',
        'Medium',
        'High',
        'Very High'
    ]
)
```

```python
df['subscription'] = np.where(
    df['y'] == 'yes',
    1,
    0
)
```

```python
df.head()
```

```python
df.to_csv("../data/bank_cleaned.csv", index= False)

print("Export Complete")
```

## Data Cleaning Summary

- Dataset contains 45,211 banking customer records.
- No missing values were observed.
- Unknown categories were retained as business-relevant information.
- Age groups, balance segments, and campaign intensity segments were engineered.
- Target variable was converted into a binary subscription indicator.
